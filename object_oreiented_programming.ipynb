```python
class BankAccount:
  accStatus= "Active"
  def __init__(self , owner,accno,balance):
    self.owner= owner
    self.accno= accno
    self.balance= balance

  def deposit(self,amount):
    if amount>0:
      self.balance +=amount
      print(f"Deposited {amount:.2f}. Balance {self.balance:.2f}")
    else:
      print("please enter valid amount")

  def withdraw(self,amount):
    if self.balance>=amount:
      self.balance -= amount
      print(f"\n withdrew {amount:.2f},. Balance:{self.balance:.2f}")
    else:
      print("Insufficient balance")
  def display (self):
    print(f" \n owner :{self.owner }\n Account no: {self.accno}  \n Balance:{self.balance}")

Acc1= BankAccount("saran",567340001,45000)
Acc1.display()
Acc1.deposit(400)
Acc1.display()
Acc1.withdraw(25000)
Acc1.display()
Acc1.deposit(-500)

```

```python
class savingsAccount(BankAccount):
  def __init__(self , owner,accno,balance, interest_rate, min_balance):
    super().__init__(owner, accno,balance)      # calling parent class
    self.interest_rate= interest_rate
    self.min_balance=min_balance

  def calculate_interest(self):
    interest =self.balance*self.interest_rate
    self.balance+=interest

  def withdraw(self,amount):
    if self.balance - amount>=self.min_balance:
      super().withdraw(amount)
    else:
      print("Insufficient Balance")

cust2= savingsAccount("Adrash",56730002,3000,0.04,1000)
cust2.display()
cust2.calculate_interest()
cust2.display()
```
