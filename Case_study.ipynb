Bank details

```python
class Bank_account:
  def __init__ (self,account_owner,account,balance):
    self.account_owner = account_owner
    self.account = account
    self. balance = balance
  def deposit (self,amount):
    if amount >0:
      self.balance += amount
      print(f"you deposited {amount}")
    else:
      print("please enter valid amount")

  def withdrawal (self,amount):
    if self.balance >= amount:
      self.balance -= amount
      print(f"You successfully withdrew {amount } \n Your blanace is {self.balance}")
    else:
      print("insufficient balance")
  def check_balance (self):
    print(f" current balance is {self.balance}")

name = input("Enter account holder name ")
acc = int(input("Enter your account number "))
balance =float(input("Enter balance"))



bk= Bank_account(name,acc ,balance)
while True:
  print("1,Deposit money")
  print("2,Withdraw money")
  print("3,View balance")
  print("4,Exit")

  choice =int(input("Enter your choice "))
  if choice == 1:
    amount = float(input("Enter amount to deposit: "))
    bk.deposit(amount)

  elif choice == 2:
      amount = float(input("Enter amount to withdraw: "))
      bk.withdrawal(amount)

  elif choice == 3:
      bk.check_balance()

  elif choice == 4:
      print("Thank you!")
      break

  else:
      print("Invalid choice")




```

Weather details

```python
class Weather:
  def __init__(self, region ,temperature ,humidity):
    self.region = region
    self.temperature = temperature
    self.humidity = humidity

  def average_temperature(self):
    return sum(self.temperature)/ len(self.temperature)

  def humidity_average(self):
    return sum(self.humidity)/ len (self.humidity)

  def climate(self):
    avg_temp=self.average_temperature()
    avg_humidity=self.humidity_average()

    if avg_temp >=30 and avg_humidity >=28:
      print("todays weather is hot and sunny day")
    elif avg_temp <=20 and  avg_humidity >=15:
      print("todays weather is cloudy ")
    else:
      print("todays weather is rainy")

  def display(self):
    print(f"Average temperature {self.average_temperature()}")
    print(f"Average humidity{self.humidity_average()}")
    print(f"Weather condition{self.climate()}")


temperature = [15,20,28,50]
humidity= [15,13,30,46]

weather = Weather("Kerala",temperature,humidity)

weather.display()


```
